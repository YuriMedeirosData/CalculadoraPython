```python
# VAMOS CRIAR UMA CALCULADORA

# ADIÇÃO +
def somar(a, b):
  somar = a + b
  return print(f"A foma dos números {a} + {b} = {somar}")

# SUBTRAÇÃO
def subtrair(a, b):
  subtrair = a - b
  return print(f"A subtração dos números {a} - {b} = {subtrair}")

# DIVISÃO /
def divisao(a, b):
  if b == 0:
    return print("ERRO: Não é possível dividir um número por zero")
  else:
    divisao = a / b
    return print(f"A divisão dos números {a} / {b} = {divisao}")

# MULTIPLICAÇÃO *
def multiplicar(a, b):
  multiplicar = a * b
  return print(f"A multiplicação dos números {a} x {b} = {multiplicar}")

while True:

  # MENU CALCULADORA PYTHON
  print("\n--- CALCULADORA PYTHON ---")
  print("[1] - SOMAR")
  print("[2] - SUBTRAIR")
  print("[3] - DIVIDIR")
  print("[4] - MULTIPLICAR")

  escolha = input("Escolha uma opção: ")

  if escolha == '0':
    print("Saindo...")
    break

  if escolha in ["1", "2", "3", "4"]:
    num1 = float(input("Digite o primeiro número: "))
    num2 = float(input("Digite o segundo número: "))
    if escolha == "1":
      resultado = somar(num1, num2)
    elif escolha == "2":
      resultado = subtrair(num1, num2)
    elif escolha == "3":
      resultado = divisao(num1, num2)
    elif escolha == "4":
      resultado = multiplicar(num1, num2)
    else:
      print("Opção indisponível")
```
